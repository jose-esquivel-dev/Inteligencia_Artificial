En esta actividad usarás el archivo de datos `Forbes_2000_top_company.csv`. Este conjunto contiene las 2000 principales empresas, clasificadas mediante el uso de cuatro métricas: ventas, ganancias, activos y valor de mercado. Tiene las 11 columnas de datos:

```python
# También se pueden enumerar los límites de todos los bins, sin necesidad de que sean equidistantes
forbes_df['Revenue (Billions)'].plot(kind='hist', bins=[0,25,50,100,200])
```

```python
# Si son equidistantes, se pueden indicar con el método arange(inicio, fin, incremento) de numpy
import numpy as np
forbes_df['Revenue (Billions)'].plot.hist(bins=np.arange(0,200,25))
```

```python
# El método hist() del dataframe, obtiene un histograma para cada columna numérica
# Se puede ajustar el tamaño con figsize
forbes_df.hist(figsize=(12,10))
```

```python
categoricas = forbes_df.select_dtypes(include='object').columns

for columna in categoricas:

    conteo = forbes_df[columna].value_counts().head(10)

    plt.figure(figsize=(10, 5))

    plt.bar(conteo.index, conteo.values)

    plt.title('Frecuencia de ' + columna)
    plt.xlabel(columna)
    plt.ylabel('Cantidad')

    plt.xticks(rotation=90)

    plt.tight_layout()
    plt.show()
```

```python
numericas = forbes_df.select_dtypes(include='number').columns

for columna in numericas:

    datos = forbes_df[columna].dropna()

    plt.figure(figsize=(6, 5))

    plt.boxplot(datos)

    plt.title('Boxplot de ' + columna)
    plt.ylabel(columna)

    plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    forbes_df['Revenue (Billions)'],
    forbes_df['Market Value (Billions)']
)

plt.title('Revenue vs Market Value')
plt.xlabel('Revenue (Billions)')
plt.ylabel('Market Value (Billions)')

plt.show()
```

```python
heatmap_df = forbes_df.copy()

categoricas_heatmap = heatmap_df.select_dtypes(include='object').columns


for columna in categoricas_heatmap:
    heatmap_df[columna] = pd.factorize(heatmap_df[columna])[0]


correlacion = heatmap_df.corr()

plt.figure(figsize=(14, 10))

plt.imshow(correlacion)

plt.colorbar()

plt.xticks(
    range(len(correlacion.columns)),
    correlacion.columns,
    rotation=90
)

plt.yticks(
    range(len(correlacion.columns)),
    correlacion.columns
)

plt.title('Mapa de calor de todas las columnas')

plt.tight_layout()
plt.show()
```
