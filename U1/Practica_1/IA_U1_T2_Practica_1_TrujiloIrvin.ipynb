Hacer la gráfica de barras, boxplot, diagramas de dispersión, mapa de calor.

Con los datos categoricos, identificar datos atipicos con boxplot, visualizar la dispersion de datos y si existe correlacion entre las columnas con el mapa de calor.

```python
# Configuración visual para que se vean mejor
sns.set_theme(style="whitegrid")

# =====================================================================
# 1. GRÁFICAS DE BARRAS (1 por cada columna categórica)
# =====================================================================
print("Generando gráficas de barras para columnas categóricas...")
# Seleccionamos automáticamente las columnas que son de tipo texto/objeto
columnas_categoricas = forbes_df.select_dtypes(include=['object']).columns

for col in columnas_categoricas:
    plt.figure(figsize=(10, 5))
    # Usamos value_counts() para contar y head(15) para no saturar la gráfica
    forbes_df[col].value_counts().head(15).plot(kind='bar', color='cornflowerblue')
    plt.title(f'Top 15 - Gráfica de Barras para: {col}')
    plt.ylabel('Cantidad')
    plt.xlabel(col)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# =====================================================================
# 2. BOXPLOTS (1 por cada columna numérica)
# =====================================================================
print("Generando boxplots para columnas numéricas...")
# Seleccionamos automáticamente las columnas que son numéricas
columnas_numericas = forbes_df.select_dtypes(include=['int64', 'float64']).columns

for col in columnas_numericas:
    plt.figure(figsize=(8, 4))
    sns.boxplot(data=forbes_df, x=col, color='lightgreen')
    plt.title(f'Boxplot de: {col}')
    plt.tight_layout()
    plt.show()

# =====================================================================
# 3. DIAGRAMAS DE DISPERSIÓN (Comparar 2 columnas con posible relación)
# =====================================================================
print("Generando diagramas de dispersión...")

# Gráfica A: Ganancias vs Valor de Mercado
plt.figure(figsize=(10, 6))
sns.scatterplot(data=forbes_df, x='Profits (Billions)', y='Market Value (Billions)', alpha=0.6, color='purple')
plt.title('Dispersión: Ganancias vs Valor de Mercado')
plt.xlabel('Ganancias (Billones)')
plt.ylabel('Valor de Mercado (Billones)')
plt.tight_layout()
plt.show()

# Gráfica B: Activos vs Valor de Mercado
plt.figure(figsize=(10, 6))
sns.scatterplot(data=forbes_df, x='Assets (Billions)', y='Market Value (Billions)', alpha=0.6, color='coral')
plt.title('Dispersión: Activos vs Valor de Mercado')
plt.xlabel('Activos (Billones)')
plt.ylabel('Valor de Mercado (Billones)')
plt.tight_layout()
plt.show()

# =====================================================================
# 4. MAPA DE CALOR (1 gráfica para correlación de todas las numéricas)
# =====================================================================
print("Generando mapa de calor...")
plt.figure(figsize=(12, 8))
# Calculamos la matriz de correlación solo para las variables numéricas
correlacion = forbes_df.corr(numeric_only=True)

# Generamos el mapa de calor
sns.heatmap(correlacion, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5, vmin=-1, vmax=1)
plt.title('Mapa de Calor de Correlaciones Numéricas')
plt.tight_layout()
plt.show()
```
