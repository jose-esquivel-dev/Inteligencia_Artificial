Hacer la gráfica de barras, boxplot, diagramas de dispersión, mapa de calor.

Con los datos categoricos, identificar datos atipicos con boxplot, visualizar la dispersion de datos y si existe correlacion entre las columnas con el mapa de calor.

```python
# Estilo
sns.set_theme(style="whitegrid")

# 1. Gráfica de barras
plt.figure(figsize=(12, 6))
forbes_df['Industry'].value_counts().head(10).plot(kind='bar', color='cornflowerblue')
plt.title('Top 10 Industrias con más empresas en la lista Forbes 2000')
plt.xlabel('Industria')
plt.ylabel('Cantidad de Empresas')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# 2. Boxplot (Identificación de datos atípicos)
plt.figure(figsize=(12, 6))
columnas_boxplot = ['Profits (Billions)', 'Market Value (Billions)', 'Revenue (Billions)']
sns.boxplot(data=forbes_df[columnas_boxplot], palette="Set2")
plt.title('Distribución y Datos Atípicos (Boxplot)')
plt.ylabel('Monto (Billones)')
plt.tight_layout()
plt.show()

# 3. Diagrama de dispersión (Visualizar dispersión de datos)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=forbes_df, x='Profits (Billions)', y='Market Value (Billions)', alpha=0.6, color='purple')
plt.title('Dispersión: Ganancias vs Valor de Mercado')
plt.xlabel('Ganancias (Billones)')
plt.ylabel('Valor de Mercado (Billones)')
plt.tight_layout()
plt.show()

# 4. Mapa de calor (Correlación entre columnas numéricas)
plt.figure(figsize=(10, 8))
correlacion = forbes_df.corr(numeric_only=True)
sns.heatmap(correlacion, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5, vmin=-1, vmax=1)
plt.title('Mapa de Calor de Correlaciones Numéricas')
plt.tight_layout()
plt.show()
```
